## 1. Compute the Factorial of a Number
Develop a function that calculates the factorial of a given number using iterative logic.
### how it work?
* Accept a number from the user.
* Validate input must be non-negative.
* Initialize result as 1.
* Multiply numbers from 1 to n.
* Return and display the factorial.

```python
# Defines a function
def factorial(n):
    if n < 0:
        return "Factorial is not defined for negative numbers."
    
    result = 1
    for i in range(1, n + 1):
        result *= i
    return result
    
# take input from users
number = int(input("Enter a number: "))

# call factorial function by passing argument and print the output by concatinatig the string
print(f"Factorial:, {factorial(number)}")
```

- Conditional statements
- Mathematical logic

## 2. Count the Number of Vowels in a String
counts vowels using string processing and conditional logic.
### how it work ?
* Receive a string input.
* Define vowels letters for look up.
* Loop through each character.
* Check whether the character is a vowel.
* Count and display total vowels.

```python
# define a function
def count_vowels(text):

    # string containing all vowels for comparison
    vowels = "aeiou"
    count = 0

    # Convert text to lowercase and check each character is a member of vowels
    for char in text.lower():
        if char in vowels:
            count += 1      
    return count

# take input from users
text = input("Enter sentence: ")

# Display the total number of vowels found in the text
print(f"Number of vowels: {count_vowels(text)}")
```

**Concepts Used**
* String iteration
* Membership operators `in`
* Functions
* Case handling

## 3. CLI-Based Calculator Using Functions
Design a command-line calculator that performs arithmetic operations using modular functions.
### Algorithm
1. Define functions for addition, subtraction, multiplication, and division.
2. Display operation menu.
3. Accept user choice and numbers.
4. Execute selected function.
5. Display result.

```python
# define a function
def calculator():

    # greet the user when the program starts
    print("Welcome to the Calculator")

    #  Keep running until the user decides to stop
    while True:

        # ask for the user to enter
        num1 = float(input("Enter first number: "))
        operation = input("Choose operation (+, -, *, /): ")
        num2 = float(input("Enter second number: "))

        # Perform the chosen arithmetic operation
        if operation == "+":
            result = num1 + num2
        elif operation == "-":
            result = num1 - num2
        elif operation == "*":
            result = num1 * num2
        elif operation == "/":
            if num2 != 0:    # Prevent division by zero
                result = num1 / num2
            else:
                print("Error: Cannot divide by zero")
                continue
        else:
            # handle invalid operation input
            print("Invalid operation")
            continue

        print("Result:", result)

        # Ask the user if they want to continue and terminate is not yes
        next_calc = input("Do another calculation? (yes/no): ").lower()
        if next_calc != "yes":
            break
```

```python
# call the function
calculator()
```

**concept used**
* Function definition
* Input/Output
* Loop
* conditional statements
* Arithmetic operations
* Error handling logic
* Control flow
